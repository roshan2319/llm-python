```python
from dotenv import load_dotenv
import os
from google import genai
```

```python
load_dotenv()

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)
```

```python
# Zero Shot Prompting
```

```python
response = client.models.generate_content(
    model="gemini-flash-latest",
    contents="Explain Machine Learning in simple words."
)

print(response.text)
```

```python
# One Shot Prompting
```

```python
prompt = """
English: Hello
Spanish: Hola

English: Thank you
Spanish:
"""

response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=prompt
)

print(response.text)
```

```python
#Template Prompting
```

```python
role = "Machine Learning Engineer"
topic = "Neural Networks"
level = "Beginner"

prompt = f"""
You are a {role}.

Explain {topic}.

Target Audience:
{level}

Rules:
- Use simple English.
- Give two examples.
- End with a summary.
"""

response = client.models.generate_content(
    model="gemini-flash-latest",
    contents=prompt
)

print(response.text)
```
